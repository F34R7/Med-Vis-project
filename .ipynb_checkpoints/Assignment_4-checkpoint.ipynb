First dataset: https://data.cdc.gov/Policy/Table-of-Gross-Cigarette-Tax-Revenue-Per-State-Orz/rkpp-igza/about_data

The first dataset has tax revenue based on tobaco articles per US State and will be an economic factor to our analysis.

Second dataset: https://data.cdc.gov/Health-Consequences-and-Costs/Smoking-Attributable-Mortality-Morbidity-and-Econo/4yyu-3s69/about_data

The second dataset has annual deaths attributed to smoking. This deaths are an estimation of deaths that could have been prevented if the patient would have not consumed tobacco.

Third dataset: https://wonder.cdc.gov/controller/datarequest/D140;jsessionid=7C8FE5C07F5B8C94557BE10A958F

The third dataset is a costum query where I looked for total deaths caused by cancer in lung and respitory systems. All datasets are for the years 2005 to 2009.

```python
import pandas
```

```python
df = pandas.read_csv("./data/Table_of_Gross_Cigarette_Tax_Revenue_Per_State_(Orzechowski_and_Walker_Tax_Burden_on_Tobacco)_20260417.csv")
df2 = pandas.read_csv("./data/Lung_Cancer_Deaths.csv")
df3 = pandas.read_csv("./data/Smoking-Attributable_Mortality,_Morbidity,_and_Economic_Costs_(SAMMEC)_-_Smoking-Attributable_Mortality_(SAM)_20260417.csv")
```

```python
df.drop(["Datasource", "Source", "TopicTypeId", "TopicId", "MeasureId"], axis = 1, inplace = True)
df_filter = df[(df["Year"] >=  2005) & (df["Year"] <= 2009)]
df_filter["Data_Value"] = pandas.to_numeric(df_filter["Data_Value"].str.replace(",", ""))
revenue_avg = df_filter.groupby(["LocationAbbr", "LocationDesc"])["Data_Value"].mean().reset_index()
```

```python
revenue_avg
```

```python
df3.drop(["DataSource", "Data_Value_Unit", "Data_Value_Footnote_Symbol", "Data_Value_Footnote", "Data_Value_Type", "TopicTypeID", "TopicID"], axis = 1, inplace = True)
df3["Data_Value"] = pandas.to_numeric(df3["Data_Value"].str.replace(",", ""), errors="coerce")
# We only want 'Overall' (both genders) and the mortality measure
df3 = df3[(df3["Sex"] == "Overall") & (df3["MeasureDesc"] == "Average Annual SAM")]
df3 = df3[["LocationAbbr", "Data_Value"]].rename(columns={"Data_Value": "Avg_Annual_SAM"})
```

```python
df3
```

```python
df2 = df2.iloc[:52]
df2 = df2.rename(columns={"State":"LocationDesc"})
df2
```

```python
final = pandas.merge(revenue_avg,df3, on = "LocationAbbr", how = "inner")
final = pandas.merge(final,df2, on = "LocationDesc", how = "inner")
final.rename(columns={"Data_Value": "Mean_State_Revenue", "Deaths":"Average_Cancer_Deaths"}, inplace = True)
```

```python
final
```

Above is the final result of the cleaned and prepared dataset, what follows is the implementation of an interactive as well as static plots and charts to visualize the data!

```python
import plotly.express as px
import plotly.io as pio
import plotly.graph_objects as go
from ipywidgets import VBox
import streamlit as st
pio.renderers.default = "notebook"


# Pre-calculate the derived metrics as new columns so they are easy to access
final["Revenue_per_SAM"] = final["Mean_State_Revenue"] / final["Avg_Annual_SAM"]
final["SAM_per_Cancer_Death"] = final["Avg_Annual_SAM"] / final["Average_Cancer_Deaths"]
final["Revenue_per_Cancer_Death"] = final["Mean_State_Revenue"] / final["Average_Cancer_Deaths"]
```

```python
import streamlit as st
import plotly.graph_objects as go
import plotly.express as px
import pandas as pd

st.title("State Revenue and Mortality Dashboard")

METRIC_MAPPING = {
    "Tax Revenue": "Mean_State_Revenue",
    "Smoking Mortality (SAM)": "Avg_Annual_SAM",
    "Revenue per SAM": "Revenue_per_SAM",
    "SAM per Respiratory Cancers Death": "SAM_per_Cancer_Death",
    "Revenue per Respiratory Cancers Death": "Revenue_per_Cancer_Death"
}

# --- 2. SESSION STATE MANAGEMENT ---
if "selected_states" not in st.session_state:
    st.session_state["selected_states"] = []
# NEW: A counter to force the map to refresh when cleared
if "map_key" not in st.session_state:
    st.session_state["map_key"] = 0

def clear_selection():
    st.session_state["selected_states"] = []
    st.session_state["map_key"] += 1 # NEW: Modifying this forces the map to drop its selections

# --- 3. CHOROPLETH MAP SECTION ---
st.header("1. National Overview")
st.markdown("*💡 **Tip:** You can click on a state on the map to add it to your selection below!*")

map_metric = st.selectbox(
    "Select a Metric to Map:",
    options=list(METRIC_MAPPING.keys())
)

fig_map = go.Figure()
fig_map.add_trace(go.Choropleth(
    locations=final["LocationDesc"],
    z=final[METRIC_MAPPING[map_metric]],
    locationmode="USA-states",
    name=map_metric,
    colorscale="Viridis"
))

fig_map.update_layout(
    title={"text": f"Average {map_metric} (2005-2009)", "x": 0.5, "xanchor": "center"},
    geo_scope='usa',
    clickmode='event+select'
)

map_event = st.plotly_chart(
    fig_map, 
    use_container_width=True, 
    on_select="rerun",
    key=f"map_{st.session_state['map_key']}" # NEW: Attach the dynamic key here
)

# NEW: Track if we actually added a new state so we can force an immediate refresh
state_added = False 

if map_event and "selection" in map_event and "points" in map_event["selection"]:
    for pt in map_event["selection"]["points"]:
        state_clicked = pt.get("location") 
        if state_clicked and state_clicked not in st.session_state["selected_states"]:
            st.session_state["selected_states"].append(state_clicked)
            state_added = True # Flag that a new state was clicked

# NEW: Force Streamlit to instantly update the multiselect below
if state_added:
    st.rerun()

st.divider()

# --- 4. STATE SELECTION & SCATTER PLOT SECTION ---
st.header("2. State Comparison & Correlation")

col_select, col_clear = st.columns([4, 1])
with col_select:
    st.multiselect(
        "Select States to analyze:",
        options=final["LocationDesc"].unique(),
        key="selected_states" 
    )
with col_clear:
    st.write("") 
    st.write("")
    st.button("Clear All States", on_click=clear_selection)

if len(st.session_state["selected_states"]) > 0:
    st.subheader("Scatter Plot Configuration")
    
    col_x, col_y = st.columns(2)
    with col_x:
        x_axis_name = st.selectbox("Select X-Axis Metric:", list(METRIC_MAPPING.keys()), index=0)
    with col_y:
        y_axis_name = st.selectbox("Select Y-Axis Metric:", list(METRIC_MAPPING.keys()), index=1)
        
    df_selected = final[final["LocationDesc"].isin(st.session_state["selected_states"])]
    
    x_col = METRIC_MAPPING[x_axis_name]
    y_col = METRIC_MAPPING[y_axis_name]
    
    if len(df_selected) >= 2:
        correlation = df_selected[x_col].corr(df_selected[y_col])
        st.write(f"**Pearson Correlation (r) for selected states:** {correlation:.3f}")
        
        fig_scatter = px.scatter(
            df_selected,
            x=x_col,
            y=y_col,
            text="LocationDesc",
            trendline="ols",
            title=f"{y_axis_name} vs. {x_axis_name}"
        )
        
        fig_scatter.update_traces(textposition='top center')
        st.plotly_chart(fig_scatter, use_container_width=True)
    else:
        st.info("Please select at least two states to view the scatter plot and regression line.")
else:
    st.info("Start by clicking states on the map or using the dropdown to view their correlation.")
```
