# Assignment 4: Smoking, Tax Revenue, and Mortality Dashboard

This notebook analyzes the relationship between **cigarette tax revenue**, **smoking-attributable mortality**, and **lung/respiratory cancer deaths** across U.S. states.

The notebook is organized in this order:

1. Project setup
2. Load source datasets
3. Clean each dataset
4. Merge into one final dataset
5. Project overview
6. Visualizations
7. Final data tables
8. Optional Streamlit dashboard export

> Important: this notebook is meant to run in **Jupyter**. The optional Streamlit app is generated at the end as a separate `app.py` file and should be run from Terminal using `streamlit run app.py`.

## 1. Project Setup

Run this cell first. It imports the required Python libraries and defines the file paths used in the project.

If any package is missing, install the requirements from Terminal:

```bash
python3 -m pip install pandas plotly statsmodels nbformat streamlit
```

````python
from pathlib import Path

import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio

pio.renderers.default = "notebook_connected"
pd.set_option("display.max_columns", None)

DATA_DIR = Path("data")

REVENUE_FILE = DATA_DIR / "Table_of_Gross_Cigarette_Tax_Revenue_Per_State_(Orzechowski_and_Walker_Tax_Burden_on_Tobacco)_20260417.csv"
CANCER_FILE = DATA_DIR / "Lung_Cancer_Deaths.csv"
SAM_FILE = DATA_DIR / "Smoking-Attributable_Mortality,_Morbidity,_and_Economic_Costs_(SAMMEC)_-_Smoking-Attributable_Mortality_(SAM)_20260417.csv"

required_files = [REVENUE_FILE, CANCER_FILE, SAM_FILE]
missing_files = [str(file) for file in required_files if not file.exists()]

if missing_files:
    print("Missing files. Make sure the data folder is in the same folder as this notebook:")
    for file in missing_files:
        print(" -", file)
else:
    print("All required data files were found.")
````

## 2. Load the Source Datasets

This project uses three datasets:

| Dataset | What it represents |
|---|---|
| Cigarette tax revenue | State cigarette tax revenue from 2005 to 2009 |
| Smoking-attributable mortality | Estimated average annual deaths attributable to smoking |
| Lung/respiratory cancer deaths | Death counts for lung and respiratory system cancers |

Only a small preview is shown here so the notebook stays readable.

````python
revenue_raw = pd.read_csv(REVENUE_FILE)
cancer_raw = pd.read_csv(CANCER_FILE)
sam_raw = pd.read_csv(SAM_FILE)

print("Revenue dataset shape:", revenue_raw.shape)
print("Cancer deaths dataset shape:", cancer_raw.shape)
print("SAM dataset shape:", sam_raw.shape)
````

````python
print("Cigarette tax revenue preview")
display(revenue_raw.head())

print("Lung/respiratory cancer deaths preview")
display(cancer_raw.head())

print("Smoking-attributable mortality preview")
display(sam_raw.head())
````

## 3. Clean the Cigarette Tax Revenue Dataset

Goal: calculate the **average cigarette tax revenue per state** for the years **2005 to 2009**.

The final cleaned table from this step is called `revenue_avg`.

````python
revenue = revenue_raw.copy()

# Remove metadata columns that are not needed for analysis.
revenue = revenue.drop(
    columns=["Datasource", "Source", "TopicTypeId", "TopicId", "MeasureId"],
    errors="ignore"
)

# Keep only the years used in this project.
revenue_2005_2009 = revenue[
    (revenue["Year"] >= 2005) &
    (revenue["Year"] <= 2009)
].copy()

# Convert revenue values from strings like "1,234" to numeric values.
revenue_2005_2009["Data_Value"] = pd.to_numeric(
    revenue_2005_2009["Data_Value"].astype(str).str.replace(",", "", regex=False),
    errors="coerce"
)

# Average revenue by state.
revenue_avg = (
    revenue_2005_2009
    .groupby(["LocationAbbr", "LocationDesc"], as_index=False)["Data_Value"]
    .mean()
    .rename(columns={"Data_Value": "Mean_State_Revenue"})
)

revenue_avg.head()
````

## 4. Clean the Smoking-Attributable Mortality Dataset

Goal: keep the **overall average annual SAM** value for each state.

SAM means **Smoking-Attributable Mortality**, which estimates deaths attributable to smoking.

The final cleaned table from this step is called `sam_avg`.

````python
sam = sam_raw.copy()

sam = sam.drop(
    columns=[
        "DataSource",
        "Data_Value_Unit",
        "Data_Value_Footnote_Symbol",
        "Data_Value_Footnote",
        "Data_Value_Type",
        "TopicTypeID",
        "TopicID"
    ],
    errors="ignore"
)

sam["Data_Value"] = pd.to_numeric(
    sam["Data_Value"].astype(str).str.replace(",", "", regex=False),
    errors="coerce"
)

sam_avg = sam[
    (sam["Sex"] == "Overall") &
    (sam["MeasureDesc"] == "Average Annual SAM")
].copy()

sam_avg = sam_avg[["LocationAbbr", "Data_Value"]].rename(
    columns={"Data_Value": "Avg_Annual_SAM"}
)

sam_avg.head()
````

## 5. Clean the Lung/Respiratory Cancer Deaths Dataset

Goal: keep the state-level death counts and rename columns so the dataset can be merged with the other two datasets.

The final cleaned table from this step is called `cancer_deaths`.

````python
cancer_deaths = cancer_raw.copy()

# The original notebook used the first 52 rows to keep state-level entries.
cancer_deaths = cancer_deaths.iloc[:52].copy()

cancer_deaths = cancer_deaths.rename(columns={"State": "LocationDesc"})

cancer_deaths["Deaths"] = pd.to_numeric(
    cancer_deaths["Deaths"].astype(str).str.replace(",", "", regex=False),
    errors="coerce"
)

cancer_deaths = cancer_deaths.rename(columns={"Deaths": "Average_Cancer_Deaths"})

cancer_deaths.head()
````

## 6. Create the Final Merged Dataset

The three cleaned datasets are merged into one table called `final`.

The final dataset includes:

| Column | Meaning |
|---|---|
| `LocationAbbr` | State abbreviation used for maps |
| `LocationDesc` | Full state name |
| `Mean_State_Revenue` | Average cigarette tax revenue from 2005 to 2009 |
| `Avg_Annual_SAM` | Average annual smoking-attributable deaths |
| `Average_Cancer_Deaths` | Lung/respiratory cancer deaths |
| `Revenue_per_SAM` | Tax revenue divided by SAM deaths |
| `SAM_per_Cancer_Death` | SAM deaths divided by cancer deaths |
| `Revenue_per_Cancer_Death` | Tax revenue divided by cancer deaths |

````python
final = pd.merge(
    revenue_avg,
    sam_avg,
    on="LocationAbbr",
    how="inner"
)

final = pd.merge(
    final,
    cancer_deaths,
    on="LocationDesc",
    how="inner"
)

# Derived metrics used in visualizations.
final["Revenue_per_SAM"] = final["Mean_State_Revenue"] / final["Avg_Annual_SAM"]
final["SAM_per_Cancer_Death"] = final["Avg_Annual_SAM"] / final["Average_Cancer_Deaths"]
final["Revenue_per_Cancer_Death"] = final["Mean_State_Revenue"] / final["Average_Cancer_Deaths"]

# Sort alphabetically for easier reading.
final = final.sort_values("LocationDesc").reset_index(drop=True)

print("Final dataset shape:", final.shape)
final.head()
````

## 7. Project Overview

This section gives a quick summary of the cleaned final dataset before the visualizations.

````python
summary_cards = pd.DataFrame({
    "Metric": [
        "States analyzed",
        "Average tax revenue",
        "Average annual SAM deaths",
        "Average lung/respiratory cancer deaths"
    ],
    "Value": [
        f"{len(final):,}",
        f"${final['Mean_State_Revenue'].mean():,.0f}",
        f"{final['Avg_Annual_SAM'].mean():,.0f}",
        f"{final['Average_Cancer_Deaths'].mean():,.0f}"
    ]
})

summary_cards
````

````python
METRIC_MAPPING = {
    "Tax Revenue": "Mean_State_Revenue",
    "Smoking Mortality (SAM)": "Avg_Annual_SAM",
    "Respiratory Cancer Deaths": "Average_Cancer_Deaths",
    "Revenue per SAM Death": "Revenue_per_SAM",
    "SAM per Respiratory Cancer Death": "SAM_per_Cancer_Death",
    "Revenue per Respiratory Cancer Death": "Revenue_per_Cancer_Death"
}

DISPLAY_LABELS = {
    "Mean_State_Revenue": "Tax Revenue",
    "Avg_Annual_SAM": "Smoking-Attributable Mortality",
    "Average_Cancer_Deaths": "Respiratory Cancer Deaths",
    "Revenue_per_SAM": "Revenue per SAM Death",
    "SAM_per_Cancer_Death": "SAM per Respiratory Cancer Death",
    "Revenue_per_Cancer_Death": "Revenue per Respiratory Cancer Death"
}
````

# 8. Visualizations

The visualizations are organized by the question they answer:

1. **Map:** where are values highest or lowest geographically?
2. **Bar charts:** which states rank highest for each metric?
3. **Scatter plot:** how do two metrics relate to each other?
4. **Correlation heatmap:** which numeric variables move together?

## 8.1 U.S. Choropleth Map

This map shows the geographic pattern for a selected metric. It uses `LocationAbbr` because Plotly's U.S. state maps expect abbreviations like `CA`, `NY`, and `TX`.

````python
def make_choropleth(metric_name="Tax Revenue"):
    column = METRIC_MAPPING[metric_name]

    fig = go.Figure(
        data=go.Choropleth(
            locations=final["LocationAbbr"],
            z=final[column],
            text=final["LocationDesc"],
            locationmode="USA-states",
            colorscale="Blues",
            colorbar_title=metric_name,
            marker_line_color="white",
            marker_line_width=0.5
        )
    )

    fig.update_layout(
        title={"text": f"{metric_name} by State", "x": 0.5, "xanchor": "center"},
        geo_scope="usa",
        template="plotly_white",
        paper_bgcolor="white",
        plot_bgcolor="white",
        font=dict(color="#111827"),
        margin=dict(l=0, r=0, t=60, b=0),
        height=550
    )

    return fig

fig_map = make_choropleth("Tax Revenue")
fig_map.show()
````

## 8.2 Top States Bar Charts

These bar charts rank states by each metric. They are easier to read than raw tables because the highest states appear clearly.

````python
def make_top_states_bar(metric_name, top_n=10):
    column = METRIC_MAPPING[metric_name]

    top_states = (
        final
        .sort_values(column, ascending=False)
        .head(top_n)
        .sort_values(column, ascending=True)
    )

    fig = px.bar(
        top_states,
        x=column,
        y="LocationDesc",
        orientation="h",
        title=f"Top {top_n} States by {metric_name}",
        labels={column: metric_name, "LocationDesc": "State"},
        hover_data=["LocationAbbr", "Mean_State_Revenue", "Avg_Annual_SAM", "Average_Cancer_Deaths"]
    )

    fig.update_traces(marker_color="#3B82F6")
    fig.update_layout(
        template="plotly_white",
        paper_bgcolor="white",
        plot_bgcolor="white",
        font=dict(color="#111827"),
        title_font_size=22,
        margin=dict(l=20, r=20, t=60, b=20),
        height=500
    )

    return fig
````

````python
fig_bar_revenue = make_top_states_bar("Tax Revenue", top_n=10)
fig_bar_revenue.show()
````

````python
fig_bar_sam = make_top_states_bar("Smoking Mortality (SAM)", top_n=10)
fig_bar_sam.show()
````

````python
fig_bar_cancer = make_top_states_bar("Respiratory Cancer Deaths", top_n=10)
fig_bar_cancer.show()
````

## 8.3 Indexed Comparison Chart

Tax revenue, SAM deaths, and cancer deaths use different units, so this chart converts each one to a **0–100 index**.

For each metric, the highest state becomes `100`, and the other states are scaled relative to it.

````python
comparison = final.copy()

index_columns = [
    "Mean_State_Revenue",
    "Avg_Annual_SAM",
    "Average_Cancer_Deaths"
]

for column in index_columns:
    comparison[f"{column}_Index"] = comparison[column] / comparison[column].max() * 100

top_comparison = (
    comparison
    .sort_values("Average_Cancer_Deaths", ascending=False)
    .head(10)
)

comparison_long = top_comparison.melt(
    id_vars=["LocationDesc"],
    value_vars=[
        "Mean_State_Revenue_Index",
        "Avg_Annual_SAM_Index",
        "Average_Cancer_Deaths_Index"
    ],
    var_name="Metric",
    value_name="Index Value"
)

comparison_long["Metric"] = comparison_long["Metric"].replace({
    "Mean_State_Revenue_Index": "Tax Revenue Index",
    "Avg_Annual_SAM_Index": "SAM Deaths Index",
    "Average_Cancer_Deaths_Index": "Cancer Deaths Index"
})

fig_indexed = px.bar(
    comparison_long,
    x="Index Value",
    y="LocationDesc",
    color="Metric",
    orientation="h",
    barmode="group",
    title="Indexed Comparison of Revenue, SAM Deaths, and Cancer Deaths",
    labels={"LocationDesc": "State", "Index Value": "Index Value (highest state = 100)"}
)

fig_indexed.update_layout(
    template="plotly_white",
    paper_bgcolor="white",
    plot_bgcolor="white",
    font=dict(color="#111827"),
    title_font_size=22,
    margin=dict(l=20, r=20, t=60, b=20),
    height=600
)

fig_indexed.show()
````

## 8.4 Scatter Plot and Correlation

This chart compares two selected metrics. A trendline is included when `statsmodels` is installed.

The code also checks that the X and Y variables are not the same. If they are the same, the result is not meaningful.

````python
def make_scatter(x_metric_name="Tax Revenue", y_metric_name="Smoking Mortality (SAM)"):
    x_col = METRIC_MAPPING[x_metric_name]
    y_col = METRIC_MAPPING[y_metric_name]

    if x_col == y_col:
        print("Please choose two different metrics for the X-axis and Y-axis.")
        return None

    correlation = final[x_col].corr(final[y_col])
    print(f"Pearson correlation between {x_metric_name} and {y_metric_name}: {correlation:.3f}")

    try:
        fig = px.scatter(
            final,
            x=x_col,
            y=y_col,
            text="LocationAbbr",
            hover_name="LocationDesc",
            trendline="ols",
            title=f"{y_metric_name} vs. {x_metric_name}",
            labels={x_col: x_metric_name, y_col: y_metric_name}
        )
    except Exception:
        fig = px.scatter(
            final,
            x=x_col,
            y=y_col,
            text="LocationAbbr",
            hover_name="LocationDesc",
            title=f"{y_metric_name} vs. {x_metric_name}",
            labels={x_col: x_metric_name, y_col: y_metric_name}
        )
        print("Trendline was skipped. Install statsmodels to enable it: python3 -m pip install statsmodels")

    fig.update_traces(textposition="top center")
    fig.update_layout(
        template="plotly_white",
        paper_bgcolor="white",
        plot_bgcolor="white",
        font=dict(color="#111827"),
        title_font_size=22,
        margin=dict(l=20, r=20, t=60, b=20),
        height=550
    )

    return fig

fig_scatter = make_scatter("Tax Revenue", "Smoking Mortality (SAM)")
fig_scatter.show()
````

## 8.5 Correlation Heatmap

The heatmap summarizes correlations between all numeric variables in the final dataset.

Values close to `1` mean a strong positive relationship. Values close to `-1` mean a strong negative relationship. Values close to `0` mean a weak linear relationship.

````python
corr_columns = list(METRIC_MAPPING.values())
corr_matrix = final[corr_columns].corr()

corr_matrix_display = corr_matrix.rename(index=DISPLAY_LABELS, columns=DISPLAY_LABELS)

fig_heatmap = px.imshow(
    corr_matrix_display,
    text_auto=".2f",
    color_continuous_scale="RdBu_r",
    zmin=-1,
    zmax=1,
    title="Correlation Heatmap of Project Metrics"
)

fig_heatmap.update_layout(
    template="plotly_white",
    paper_bgcolor="white",
    plot_bgcolor="white",
    font=dict(color="#111827"),
    title_font_size=22,
    margin=dict(l=20, r=20, t=60, b=20),
    height=650
)

fig_heatmap.show()
````
